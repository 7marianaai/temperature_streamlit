```python
import pandas as pd
import numpy as np

# Реальные средние температуры (примерные данные) для городов по сезонам
seasonal_temperatures = {
    "New York": {"winter": 0, "spring": 10, "summer": 25, "autumn": 15},
    "London": {"winter": 5, "spring": 11, "summer": 18, "autumn": 12},
    "Paris": {"winter": 4, "spring": 12, "summer": 20, "autumn": 13},
    "Tokyo": {"winter": 6, "spring": 15, "summer": 27, "autumn": 18},
    "Moscow": {"winter": -10, "spring": 5, "summer": 18, "autumn": 8},
    "Sydney": {"winter": 12, "spring": 18, "summer": 25, "autumn": 20},
    "Berlin": {"winter": 0, "spring": 10, "summer": 20, "autumn": 11},
    "Beijing": {"winter": -2, "spring": 13, "summer": 27, "autumn": 16},
    "Rio de Janeiro": {"winter": 20, "spring": 25, "summer": 30, "autumn": 25},
    "Dubai": {"winter": 20, "spring": 30, "summer": 40, "autumn": 30},
    "Los Angeles": {"winter": 15, "spring": 18, "summer": 25, "autumn": 20},
    "Singapore": {"winter": 27, "spring": 28, "summer": 28, "autumn": 27},
    "Mumbai": {"winter": 25, "spring": 30, "summer": 35, "autumn": 30},
    "Cairo": {"winter": 15, "spring": 25, "summer": 35, "autumn": 25},
    "Mexico City": {"winter": 12, "spring": 18, "summer": 20, "autumn": 15},
}

# Сопоставление месяцев с сезонами
month_to_season = {12: "winter", 1: "winter", 2: "winter",
                   3: "spring", 4: "spring", 5: "spring",
                   6: "summer", 7: "summer", 8: "summer",
                   9: "autumn", 10: "autumn", 11: "autumn"}

# Генерация данных о температуре
def generate_realistic_temperature_data(cities, num_years=10):
    dates = pd.date_range(start="2010-01-01", periods=365 * num_years, freq="D")
    data = []

    for city in cities:
        for date in dates:
            season = month_to_season[date.month]
            mean_temp = seasonal_temperatures[city][season]
            # Добавляем случайное отклонение
            temperature = np.random.normal(loc=mean_temp, scale=5)
            data.append({"city": city, "timestamp": date, "temperature": temperature})

    df = pd.DataFrame(data)
    df['season'] = df['timestamp'].dt.month.map(lambda x: month_to_season[x])
    return df

# Генерация данных
data = generate_realistic_temperature_data(list(seasonal_temperatures.keys()))
data.to_csv('temperature_data.csv', index=False)
```

### 2. Скользящее среднее и стандартное отклонение за 30 дней


```python
def roll_stat_get(city_data):
    res = city_data.sort_values('timestamp').copy()
    res['rolling_mean'] = res['temperature'].rolling(window=30).mean()
    res['rolling_std'] = res['temperature'].rolling(window=30).std()
    return res
```

### 3. Средняя температура и стандартное отклонение по сезонам и городам


```python
def season_stat_get(data):
    res = data.groupby(['city', 'season'])['temperature'].agg(['mean', 'std'])
    res = res.reset_index()
    res = res.rename(columns={'mean': 'season_mean', 'std': 'season_std'})
    return res
```

### 4. Аномалии относительно сезонной нормы и скользящего среднего


```python
def season_anom_add(data, stats):
    res = data.merge(stats, on=['city', 'season'])
    res['season_lower'] = res['season_mean'] - 2 * res['season_std']
    res['season_upper'] = res['season_mean'] + 2 * res['season_std']
    res['season_anomaly'] = (
        (res['temperature'] < res['season_lower'])
        | (res['temperature'] > res['season_upper'])
    )
    return res
```

```python
def roll_anom_add(data):
    res = data.copy()
    res['rolling_lower'] = res['rolling_mean'] - 2 * res['rolling_std']
    res['rolling_upper'] = res['rolling_mean'] + 2 * res['rolling_std']
    res['rolling_anomaly'] = (
        (res['temperature'] < res['rolling_lower'])
        | (res['temperature'] > res['rolling_upper'])
    )
    return res
```

### 5. Последовательный и параллельный анализ всех городов: сравнение


```python
def analyze_city(city_data):
    res = roll_stat_get(city_data)
    stats = season_stat_get(city_data)
    res = season_anom_add(res, stats)
    res = roll_anom_add(res)
    return res
```

```python
def analyze_sequent(grps):
    res = []

    for city_data in grps:
        res.append(analyze_city(city_data))

    return pd.concat(res, ignore_index=True)
```

```python
def analyze_parallel(grps):
    res = Parallel(n_jobs=2)(delayed(analyze_city)(grp) for grp in grps)
    return pd.concat(res, ignore_index=True)
```

```python
def measure(function, grps):
    start_time = time.time()
    res = function(grps)
    diff = time.time() - start_time
    return res, diff
```

```python
measurements = []

for repeat in range(10):
    if repeat % 2 == 0:
        sequent_res, sequent_time = measure(analyze_sequent, city_grps)
        parallel_res, parallel_time = measure(analyze_parallel, city_grps)
    else:
        parallel_res, parallel_time = measure(analyze_parallel, city_grps)
        sequent_res, sequent_time = measure(analyze_sequent, city_grps)

    pd.testing.assert_frame_equal(sequent_res, parallel_res)
    measurements.append({
        'Повтор': repeat + 1,
        'Последовательно, с': sequent_time,
        'Параллельно, с': parallel_time,
    })
```

```python
timing_table = pd.DataFrame(measurements)
timing_table.round(4)
```

```python
pd.testing.assert_frame_equal(analysis_res, sequent_res)
print('Результаты всех видов анализов совпадают')
```

```python
sequent_median = timing_table['Последовательно, с'].median()
parallel_median = timing_table['Параллельно, с'].median()
speedup = sequent_median / parallel_median
```

### 6. Долгосрочные тренды изменения температуры

```python
import plotly.graph_objects as go
```

```python
trend_data = temp_data.copy()
trend_data['year'] = trend_data['timestamp'].dt.year
```

```python
annual_temp = (
    trend_data
    .groupby(['city', 'year'])['temperature']
    .mean()
    .reset_index()
)

annual_temp = annual_temp.rename(
    columns={'temperature': 'year_mean'}
)

annual_temp.head(15)
```

```python
def trend_stat_get(data):
    res = []

    for city, city_data in data.groupby('city'):
        slope, intercept = np.polyfit(
            city_data['year'],
            city_data['year_mean'],
            1
        )

        period = city_data['year'].max() - city_data['year'].min()
        period_change = slope * period

        if slope > 0:
            direction = 'повышение'
        elif slope < 0:
            direction = 'снижение'
        else:
            direction = 'без изменений'

        res.append({
            'city': city,
            'trend_per_year': slope,
            'change_for_period': period_change,
            'direction': direction,
        })

    return pd.DataFrame(res)
```

```python
trend_statistics = trend_stat_get(annual_temp)

trend_statistics.round(3)
```

```python
def trend_plot(sel_city):
    city_data = annual_temp[
        annual_temp['city'] == sel_city
    ].copy()

    slope, intercept = np.polyfit(
        city_data['year'],
        city_data['year_mean'],
        1
    )

    city_data['trend_line'] = (
        slope * city_data['year'] + intercept
    )

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=city_data['year'],
            y=city_data['year_mean'],
            mode='lines+markers',
            name='Средняя температура'
        )
    )

    fig.add_trace(
        go.Scatter(
            x=city_data['year'],
            y=city_data['trend_line'],
            mode='lines',
            name='Линия тренда'
        )
    )

    fig.update_layout(
        title=f'Долгосрочный тренд температуры: {sel_city}',
        xaxis_title='Год',
        yaxis_title='Средняя температура, °C',
        hovermode='x unified'
    )

    return fig
```

### 2. Получение текущей температуры (синхронно)

```python
def weather_sync_get(city, api_key):
    params = {
        'q': city,
        'appid': api_key,
        'units': 'metric',
    }
    response = requests.get(weather_url, params=params, timeout=10)
    weather = response.json()

    if response.status_code != 200:
        message = weather.get('message', 'неизвестная ошибка')
        raise ValueError(f'Ошибка API {response.status_code}: {message}')

    local_time = pd.to_datetime(
        weather['dt'] + weather['timezone'], unit='s'
    )

    return {
        'city': city,
        'local_time': local_time,
        'current_temp': weather['main']['temp'],
    }
```

```python
def weather_now(sel_city):
    weather_sync = weather_sync_get(sel_city, api_key)
    return weather_sync
```

```python
print(weather_now('Moscow'))
print(weather_now('Tokyo'))
weather_now('Madrid')
```

### 3. Исторический нормальный диапазон текущего сезона

```python
def season_get(timestamp):
    return month_to_season[timestamp.month]
```

```python
def current_temp_compare(weather, stats):
    city = weather['city']
    season = season_get(weather['local_time'])

    city_stat = stats[(stats['city'] == city) & (stats['season'] == season)].iloc[0]

    lower = city_stat['season_mean'] - 2 * city_stat['season_std']
    upper = city_stat['season_mean'] + 2 * city_stat['season_std']
    normal = lower <= weather['current_temp'] <= upper

    return {
        'city': city,
        'local_time': weather['local_time'],
        'season': season,
        'current_temp': weather['current_temp'],
        'season_mean': city_stat['season_mean'],
        'season_lower': lower,
        'season_upper': upper,
        'status': 'нормальная' if normal else 'аномальная',
    }
```

```python
def cities_weather_check(sel_cities):
    res = []

    for city in sel_cities:
        weather = weather_sync_get(city, api_key)
        city_res = current_temp_compare(
            weather,
            season_statistics
        )
        res.append(city_res)

    return pd.DataFrame(res).round(2)
```

```python
all_cities = [
    'New York',
    'London',
    'Paris',
    'Tokyo',
    'Moscow',
    'Sydney',
    'Berlin',
    'Beijing',
    'Rio de Janeiro',
    'Dubai',
    'Los Angeles',
    'Singapore',
    'Mumbai',
    'Cairo',
    'Mexico City'
]

current_temp_table = cities_weather_check(all_cities)
current_temp_table
```

- нормальная ли или аномальная температура зависит от времени запуска, предполагаемые аномальные значения по заданию не всегда оказываются аномальными, как сейчас (живем в эпоху изменения климата))

### 4. Получение текущей температуры (асинхронно)

```python
async def weather_async_get(session, city, api_key):
    params = {
        'q': city,
        'appid': api_key,
        'units': 'metric',
    }

    async with session.get(weather_url, params=params) as response:
        weather = await response.json()

        if response.status != 200:
            message = weather.get('message', 'неизвестная ошибка')
            raise ValueError(f'Ошибка API {response.status}: {message}')

        local_time = pd.to_datetime(
            weather['dt'] + weather['timezone'], unit='s'
        )

        return {
            'city': city,
            'local_time': local_time,
            'current_temp': weather['main']['temp'],
        }
```

```python
async def weather_many_async_get(cities, api_key):
    timeout = aiohttp.ClientTimeout(total=10)

    async with aiohttp.ClientSession(timeout=timeout) as session:
        tasks = [
            weather_async_get(session, city, api_key)
            for city in cities
        ]
        return await asyncio.gather(*tasks)
```

```python
start_time = time.time()
async_weather = await weather_many_async_get(test_cities, api_key)
async_time = time.time() - start_time

print(f'Время асинхронных запросов: {async_time:.4f} с')
```

```python
async_current_res = []

for weather in async_weather:
    async_current_res.append(
        current_temp_compare(weather, season_statistics)
    )

current_temp_table = pd.DataFrame(async_current_res)
current_temp_table.round(2)
```

Результат зависит от фактической погоды в момент запуска, поэтому города, указанные в условии как нормальные или аномальные, могут получить другой статус при новом запуске, в этом запуске не было аномальных температур

### 5. Сравнение

```python
request_timing = pd.DataFrame({
    'Способ': ['Синхронный', 'Асинхронный'],
    'Время, с': [sync_time, async_time],
})
request_timing.round(4)
```

```python
if async_time < sync_time:
    request_speedup = sync_time / async_time
    print(
        f'Асинхронные запросы оказались быстрее в {request_speedup:.2f} раза.'
    )
else:
    print('В этом запуске асинхронный способ не дал ускорения.')
```

Выводы

- Для одного города удобнее синхронный способ, так как его код проще, а от асинхронности почти нет преимуществ
- Для нескольких городов лучше подходит асинхронный способ: запросы выполняются одновременно во время ожидания ответов от API
- Синхронное получение погоды для семи городов заняло 1.0481 секунды, асинхронное — 0.1639 секунды, в этом запуске асинхронный способ оказался примерно в 6.39 раза быстрее
